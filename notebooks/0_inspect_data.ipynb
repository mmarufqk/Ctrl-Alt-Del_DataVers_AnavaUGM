```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data")
PREVIEW_DIR = DATA_DIR / "preview"

PREVIEW_DIR.mkdir(exist_ok=True)
```

```python
# Paths
train_path = DATA_DIR / "train.csv"
test_path  = DATA_DIR / "test.csv"

# Read only first 10 rows (AMAN untuk file besar)
train_head = pd.read_csv(train_path, nrows=10)
test_head  = pd.read_csv(test_path, nrows=10)

# Save to CSV
train_head.to_csv(PREVIEW_DIR / "train_head_10.csv", index=False)
test_head.to_csv(PREVIEW_DIR / "test_head_10.csv", index=False)
train_head, test_head
```

```python
def show_info(df, name):
    print(f"\n{name}")
    print("-" * len(name))
    print(df.dtypes)

show_info(train_head, "TRAIN COLUMNS")
show_info(test_head, "TEST COLUMNS")
```
